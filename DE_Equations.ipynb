```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = sp.symbols('x')
y = sp.Function('y')
```

```python
#Example 1: y' = x^2
eq1 = sp.Eq(y(x).diff(x),x**2)
sol1 = sp.dsolve(eq1,y(x))
print(sol1)
```

```python
def solve_and_plot(rhs_expr, ic_value=0, xrange = (-5,5)):
    eq = sp.Eq(y(x).diff(x),rhs_expr)
    sol = sp.dsolve(eq,y(x),ics={y(0):ic_value})
    f = sp.lambdify(x,sol.rhs,'numpy')
    xs = np.linspace(*xrange,400)
    plt.plot(xs,f(xs))
    plt.title(f"y' = {rhs_expr} , y(0)={ic_value}")
    plt.grid(True);plt.show()
    return sol
    
    
```

```python
solve_and_plot(sp.sin(3*x))
```

```python
solve_and_plot(sp.sin(3*x),ic_value=2)
```
